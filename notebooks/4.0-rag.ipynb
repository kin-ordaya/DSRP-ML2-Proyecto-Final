# 4.0 - GenAI RAG: Preguntas sobre el Dataset Telco Churn

Este notebook implementa un sistema de **Retrieval-Augmented Generation (RAG)** que permite hacer preguntas en lenguaje natural sobre el dataset de churn de telecomunicaciones, usando un modelo de embeddings local y un generador local (TinyLlama).

- **Embeddings**: `sentence-transformers/all-MiniLM-L6-v2` (vectoriza los documentos).
- **Vector store**: ChromaDB (almacena y recupera los fragmentos relevantes).
- **Generador**: `TinyLlama/TinyLlama-1.1B-Chat-v1.0` (modelo local, sin API key).

```python
from src.genai.rag import setup_rag
```

### 1. Construir o cargar el pipeline RAG

La primera ejecución descarga los modelos (embeddings + TinyLlama) y construye el índice vectorial en `models/rag_chroma/`. Ejecuciones posteriores reutilizan el índice ya persistido.

```python
# rebuild=True re-indexa los documentos desde cero (útil si cambian).
# La primera vez puede tardar unos minutos en descargar los modelos.
answer = setup_rag(rebuild=False)
```

### 2. Consultas de ejemplo

El sistema recupera el contexto más relevante del dataset y genera una respuesta fundamentada en ese contexto.

```python
preguntas = [
    "Cual es la tasa de abandono del dataset?",
    "Cuales son los tipos de contrato y cual tiende a churn mas?",
    "Que variables describen los servicios contratados?",
]
for q in preguntas:
    res = answer(q)
    print(f"Q: {q}")
    print(f"A: {res['answer']}")
    print("-" * 60)
```

### 3. Inspección del contexto recuperado

Podemos ver qué fragmentos (sources) se usaron como contexto para una pregunta concreta.

```python
res = answer("Que metodo de pago es el mas comun?")
print("Respuesta:")
print(res["answer"])
print("\n" + "=" * 60)
print("Contexto recuperado (fragmentos):")
for i, src in enumerate(res["sources"], 1):
    print(f"\n--- Fragmento {i} ---")
    print(src[:300])
```

### 4. Conclusión

El sistema RAG permite consultar el dataset de forma conversacional y online, recuperando el contexto más relevante y generando respuestas fundamentadas con un modelo local (sin depender de APIs externas). Este componente GenAI complementa el modelo predictivo de churn del proyecto.
